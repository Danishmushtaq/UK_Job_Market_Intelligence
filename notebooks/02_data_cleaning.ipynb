```python
orders["delivery_delay_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.days
```

```python
orders[[
    "order_id",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
    "delivery_delay_days"
]].head(10)
```

```python
orders["delivery_delay_days"].describe()
```

```python
print("Early:", (orders["delivery_delay_days"] < 0).sum())
print("On time:", (orders["delivery_delay_days"] == 0).sum())
print("Late:", (orders["delivery_delay_days"] > 0).sum())
```

```python
total_with_delay = orders["delivery_delay_days"].notna().sum()

early_pct = (orders["delivery_delay_days"] < 0).sum() / total_with_delay * 100
on_time_pct = (orders["delivery_delay_days"] == 0).sum() / total_with_delay * 100
late_pct = (orders["delivery_delay_days"] > 0).sum() / total_with_delay * 100

print(f"Early: {early_pct:.2f}%")
print(f"On time: {on_time_pct:.2f}%")
print(f"Late: {late_pct:.2f}%")
```

### Delivery Performance

We calculated delivery delay by comparing the actual customer delivery date
with the estimated delivery date.

Among orders with both dates available:
- 91.89% arrived early
- 1.34% arrived on the estimated date
- 6.77% arrived late

Missing delivery dates were retained rather than artificially filled.
